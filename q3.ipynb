```python
import pandas as pd

# Load Excel file
df = pd.read_excel("q-excel-correlation-heatmap.xlsx")

# Show first few rows
df.head()
```

```python
# Compute correlation
corr = df.corr()

# Save correlation matrix to CSV
corr.to_csv("correlation.csv")
corr
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(6,6))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="RdYlGn",  # Red-Yellow-Green palette (close to Excel style)
    center=0,
    square=True,
    cbar=True
)
plt.title("Supply Chain Correlation Heatmap", fontsize=14)

# Save screenshot equivalent
plt.savefig("heatmap.png", dpi=150, bbox_inches="tight")
plt.show()

```
